```python
import pandas as pd
```

```python
# data presetting

df = pd.read_csv("anime.csv")
pd.set_option("display.max_colwidth",None)
df.to_string(index=False)

df
```

```python
# Adding anime name
def Real_name(x):
    box = ""
    flg = True
    for i in x:
        if i == "(":
            break
        if flg == True:
            box += i 
    return box

df["Anime_name"] = df["Title"].apply(Real_name)

df
```

```python
#make a new column for episode count
```

```python
# Adding episodes number

def Fetch(x):
    data = ""
    check = False
    for i in x:
        if i == "(":
            check = True
            continue
            
        if i ==")":
            check = False
            break

        if check==True:
            data+=i
    return data

df["Episodes"] = df["Title"].apply(Fetch).str.removesuffix("eps").astype(int)

df
```

```python
#make a new column for time stamp
```

```python
# Adding the date number

def Real_date(f):
    c= ""
    flg1 = False
    for i in f :
        if i ==")":
            flg1 = True
            continue

        if len(c) == 19:
            flg1 = False
            break

        if flg1 == True:
            c+= i
    return c

df["Anime_date"] = df["Title"].apply(Real_date)

df
```

```python
# what is the price of each anime's manga
```

```python
def Manga_prices(x):
      
      price_str = None
      if "Preview" in x:
            contain = x.split("€")
            price_str = contain[1].removesuffix("Preview")
      return price_str

df["Manga price"] = pd.to_numeric(df["Title"].apply(Manga_prices), errors ="coerce")
```

```python
#which is the longest running anime
```

```python
split_time = df["Anime_date"].str.split('-', expand=True)
start_time = split_time[0].str.strip()
end_time = split_time[1].str.strip()

start_month = pd.to_datetime(start_time, format='%b %Y', errors='coerce')
end_month =  pd.to_datetime(end_time, format='%b %Y', errors='coerce')

df["month"] = (end_month.dt.month - start_month.dt.month) + (12*(end_month.dt.year-start_month.dt.year))
```

```python
#make a column for member count 
```

```python
def Members_number(x):

    split_1 = x.rsplit(' - ', maxsplit = 1)
    temp_str = split_1[1]

    if "membersManga" in x:
        temp_str1 = temp_str.split(' membersManga ')
        temp_str2 = temp_str1[0]

    else:
        temp_str1 = temp_str.split(' members ')
        temp_str2 = temp_str1[0]

    final_str = temp_str2[8:].removesuffix(' members')
    last_str = final_str.replace(',','')
    return last_str

df["Member_count"] = df["Title"].apply(Members_number).astype(int)
```
